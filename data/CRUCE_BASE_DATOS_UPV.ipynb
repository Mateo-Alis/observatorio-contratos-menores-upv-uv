MERGE

```python
"""
Cruza los contratos menores de la UPV y la UV por PROVEEDOR + AÑO (una fila por proveedor-año)
SIN PERDER las columnas propias de cada base:
    UV  : Objeto, Tipo contrato
    UPV : CONCEPTO ECONÓMICO, CENTRO DIRECTIVO

Salidas:
  1) RESUMEN proveedor-año  -> una fila por proveedor y año. Las columnas de texto
     (Objeto, Tipo contrato, CONCEPTO ECONÓMICO, CENTRO DIRECTIVO) recogen los valores
     distintos de ese proveedor-año separados por ' | ', con '(×n)' si se repiten n veces.
  2) DETALLE de contratos   -> una fila por contrato original (UPV + UV apilados) con TODAS
     las columnas originales + PROVEEDOR_CLAVE. Sirve para ver qué Objeto / Concepto /
     Centro corresponde a cada importe concreto (en el resumen eso se pierde al agregar).

Unificación de nombres de proveedor en 3 capas:
  1. Automática y segura: nombre normalizado, SIN espacios, y con palabras ORDENADAS
     (cubre 'PERKIN ELMER' = 'PERKINELMER' y 'DOLZ CASANA JAIME' = 'JAIME DOLZ CASANA').
  2. Automática por NIF: nombres distintos de la UV con el mismo NIF son el mismo proveedor.
  3. Revisión manual: los candidatos por similitud difusa se escriben en
     'candidatos_revisar.xlsx'. Marca S en la columna MISMO_SN y vuelve a ejecutar el script.

Uso:
    pip install pandas openpyxl rapidfuzz            (opcional: xlsxwriter, escribe el Excel mucho más rápido)
    python merge_proveedor_año_v3.py      (1ª vez: genera el merge + el fichero de candidatos)
    ... revisa candidatos_revisar.xlsx (columna MISMO_SN = S / N) ...
    python merge_proveedor_año_v3.py      (2ª vez: aplica tus decisiones)
"""
import html
import os
import re
import unicodedata
from functools import lru_cache

import pandas as pd
from rapidfuzz import fuzz, process

# ---------------- CONFIG ----------------
RUTA_UPV = "CONTRATOS_MENORES_UPV_2019_2023.csv"
RUTA_UV = "CONTRATOS_MENORES_UV_2019_2023.csv"
SALIDA_CSV = "proveedor_año_UPV_UV.csv"
SALIDA_DETALLE_CSV = "contratos_detalle_UPV_UV.csv"   # None si no lo quieres
SALIDA_XLSX = "proveedor_año_UPV_UV.xlsx"             # None si no lo quieres (hojas: resumen + detalle)
FICHERO_REVISION = "candidatos_revisar.xlsx"
UMBRAL_FUZZY = 90                                     # 0-100; más alto = menos candidatos y más fiables
CANDIDATOS_POR_NOMBRE = 3
SEP = " | "                                           # separador de valores en las columnas de texto
MAX_CHARS_CELDA = 32000                               # Excel admite 32.767 caracteres por celda

EXTRAS_UPV = ["CENTRO DIRECTIVO", "CONCEPTO ECONÓMICO"]
EXTRAS_UV = ["Tipo contrato", "Objeto"]
IMPORTES_UV = ["Importe adjudicación (IVA excluido)", "Importe IVA",
               "Importe adjudicación (IVA incluido)", "Importe final contrato (IVA excluido)"]
ORDEN_DETALLE = [
    "UNIVERSIDAD", "PROVEEDOR_CLAVE", "AÑO", "Identificador / EXPEDIENTE",
    "Nombre adjudicatario / TERCERO", "NIF adjudicatario",
    "CENTRO DIRECTIVO", "Unidad funcional", "CONCEPTO ECONÓMICO", "Objeto", "Tipo contrato",
    "IMPORTE", "Importe adjudicación (IVA excluido)", "Tipo IVA", "Importe IVA",
    "Importe adjudicación (IVA incluido)", "Importe final contrato (IVA excluido)",
    "Fecha adjudicación", "Plazo ejecución", "Exclusividad",
    "¿Se trata de acceso a BBDD o publicaciones?", "DA 54 LCSP", "Pagado por caja fija",
]
# ----------------------------------------


# ---------- utilidades ----------
def leer(ruta):
    """Lee el CSV como texto. Solo la cadena vacía cuenta como nulo (así 'NA' no se pierde).
    Recorta espacios y sustituye saltos de línea internos (hay en 'Objeto') por un espacio."""
    df = pd.read_csv(ruta, sep=";", encoding="utf-8-sig", dtype=str,
                     keep_default_na=False, na_values=[""])
    for c in df.columns:
        s = df[c].str.replace(r"[\r\n]+", " ", regex=True).str.strip()
        df[c] = s.where(s != "")
    return df


def a_numero(serie):
    """Convierte importes con formatos mezclados: '1.088,52', '83,77', '199.0', '818.57'."""
    def conv(x):
        if x is None or pd.isna(x):
            return None
        x = str(x).strip()
        if x == "":
            return None
        if "," in x:
            x = x.replace(".", "").replace(",", ".")
        try:
            return float(x)
        except ValueError:
            return None
    return serie.map(conv).astype("float64")


def a_fecha(serie):
    """La UV mezcla 'AAAA-MM-DD hh:mm:ss' y números de serie de Excel (p.ej. 44473)."""
    es_serie = serie.str.fullmatch(r"\d{5}", na=False)
    fecha = pd.to_datetime(serie.where(~es_serie), format="%Y-%m-%d %H:%M:%S", errors="coerce")
    if es_serie.any():
        fecha.loc[es_serie] = pd.to_datetime(serie[es_serie].astype(int), unit="D",
                                             origin="1899-12-30")
    return fecha


def arreglar_codificacion(texto):
    """'ESPAÃ&#145;A' -> 'ESPAÑA' (entidades HTML y UTF-8 mal decodificado)."""
    t = html.unescape(str(texto))
    if "Ã" in t or "Â" in t:
        try:
            t = t.encode("latin-1").decode("utf-8")
        except (UnicodeEncodeError, UnicodeDecodeError):
            pass
    return t


@lru_cache(maxsize=None)
def normalizar_nombre(nombre):
    if pd.isna(nombre):
        return None
    base = unicodedata.normalize("NFKD", arreglar_codificacion(nombre).upper())
    base = "".join(c for c in base if not unicodedata.combining(c))
    base = re.sub(r"[^A-Z0-9 ]", " ", base)
    s = re.sub(r"\b(S ?L ?U?|S ?A|S ?L ?L|SCV|C ?B|S ?COOP)\b", " ", base)
    s = re.sub(r"\s+", " ", s).strip()
    if not s:  # el nombre era solo una forma societaria: no lo dejamos vacío
        s = re.sub(r"\s+", " ", base).strip()
    return s or None


def moda(s):
    s = s.dropna()
    return s.mode().iat[0] if len(s) else None


def unicos(s):
    """Valores distintos (más frecuentes primero) unidos por SEP; 'valor (×n)' si se repite n veces.
    Si no cabe en una celda de Excel se corta e indica cuántos valores quedan fuera."""
    s = s.dropna()
    if s.empty:
        return None
    vc = s.value_counts().sort_index(kind="stable").sort_values(ascending=False, kind="stable")
    partes = [(v if n == 1 else f"{v} (×{n})")[:MAX_CHARS_CELDA] for v, n in vc.items()]
    acum, largo = [], 0
    for p in partes:
        if largo + len(p) + len(SEP) > MAX_CHARS_CELDA:
            acum.append(f"[+{len(partes) - len(acum)} más; ver hoja detalle]")
            break
        acum.append(p)
        largo += len(p) + len(SEP)
    return SEP.join(acum)


class UnionFind:
    def __init__(self, items):
        self.p = {i: i for i in items}

    def find(self, x):
        while self.p[x] != x:
            self.p[x] = self.p[self.p[x]]
            x = self.p[x]
        return x

    def union(self, a, b):
        if a in self.p and b in self.p:
            ra, rb = self.find(a), self.find(b)
            if ra != rb:
                self.p[rb] = ra


def preparar(df, col_prov, col_importe, col_nif=None, extras=()):
    out = pd.DataFrame({
        "AÑO": pd.to_numeric(df["AÑO"], errors="coerce").astype("Int64"),
        "PROVEEDOR": df[col_prov],
        "IMPORTE": a_numero(df[col_importe]),
        "NIF": df[col_nif].str.upper() if col_nif else None,
    })
    for c in extras:                      # columnas propias de cada base
        out[c] = df[c]
    out["NORM"] = out["PROVEEDOR"].map(normalizar_nombre)
    return out[out["NORM"].notna()].copy()


# ---------- carga ----------
raw_upv = leer(RUTA_UPV)
raw_uv = leer(RUTA_UV)
upv = preparar(raw_upv, "TERCERO", "IMPORTE", extras=EXTRAS_UPV)
uv = preparar(raw_uv, "Nombre adjudicatario", "Importe adjudicación (IVA excluido)",
              col_nif="NIF adjudicatario", extras=EXTRAS_UV)
sin_nombre = (len(raw_upv) - len(upv)) + (len(raw_uv) - len(uv))
if sin_nombre:
    print(f"AVISO: {sin_nombre:,} contratos sin nombre de proveedor: no entran en el resumen, "
          f"pero sí en el detalle.")

nombres = sorted(set(upv["NORM"]) | set(uv["NORM"]))
uf = UnionFind(nombres)


def comunes():
    ga = {uf.find(n) for n in upv["NORM"].unique()}
    gb = {uf.find(n) for n in uv["NORM"].unique()}
    return len(ga & gb)


print(f"Proveedores en común (nombre normalizado exacto): {comunes():,}")

# ---------- capa 1: claves seguras (sin espacios / palabras ordenadas) ----------
for clave in (lambda n: n.replace(" ", ""), lambda n: " ".join(sorted(n.split()))):
    cubo = {}
    for n in nombres:
        k = clave(n)
        if k in cubo:
            uf.union(cubo[k], n)
        else:
            cubo[k] = n
print(f"  + sin espacios / palabras ordenadas:            {comunes():,}")

# ---------- capa 2: mismo NIF (solo UV lo tiene) ----------
# Solo NIF completos de empresa (p.ej. B96159801). Los de personas físicas vienen
# enmascarados (053*****1*) y personas distintas comparten máscara: NO se usan.
nif_ok = uv[uv["NIF"].fillna("").str.fullmatch(r"[A-Z]\d{7}[A-Z0-9]")]
for _, g in nif_ok.groupby("NIF"):
    ns = g["NORM"].unique()
    for n in ns[1:]:
        uf.union(ns[0], n)
print(f"  + mismo NIF dentro de la UV:                    {comunes():,}")


# ---------- capa 3: revisión manual ----------
def resumen_por_grupo(df, prefijo):
    d = df.assign(GRUPO=df["NORM"].map(uf.find))
    g = d.groupby("GRUPO").agg(
        norm=("NORM", moda), nombre=("PROVEEDOR", moda),
        n=("IMPORTE", "size"), importe=("IMPORTE", "sum"),
        nif=("NIF", moda),
    )
    g.columns = [f"{prefijo}_{c}" for c in g.columns]
    return g


if os.path.exists(FICHERO_REVISION):
    rev = pd.read_excel(FICHERO_REVISION, dtype=str)
    si = rev[rev["MISMO_SN"].fillna("").str.strip().str.upper() == "S"]
    for _, r in si.iterrows():
        uf.union(r["UPV_NORM"], r["UV_NORM"])
    print(f"  + {len(si):,} decisiones 'S' de {FICHERO_REVISION}:".ljust(52), f"{comunes():,}")
else:
    ra, rb = resumen_por_grupo(upv, "UPV"), resumen_por_grupo(uv, "UV")
    solo_a = ra[~ra.index.isin(rb.index)]
    solo_b = rb[~rb.index.isin(ra.index)]
    lista_b = solo_b["UV_norm"].tolist()
    idx_b = solo_b.reset_index(drop=True)
    filas = []
    for _, a in solo_a.iterrows():
        for _, score, i in process.extract(a["UPV_norm"], lista_b, scorer=fuzz.token_sort_ratio,
                                           score_cutoff=UMBRAL_FUZZY, limit=CANDIDATOS_POR_NOMBRE):
            b = idx_b.iloc[i]
            filas.append({
                "SIMILITUD": round(score), "MISMO_SN": "",
                "UPV_nombre": a["UPV_nombre"], "UPV_contratos": a["UPV_n"],
                "UPV_importe": round(a["UPV_importe"], 2),
                "UV_nombre": b["UV_nombre"], "UV_nif": b["UV_nif"],
                "UV_contratos": b["UV_n"], "UV_importe": round(b["UV_importe"], 2),
                "UPV_NORM": a["UPV_norm"], "UV_NORM": b["UV_norm"],
            })
    cand = pd.DataFrame(filas, columns=[
        "SIMILITUD", "MISMO_SN", "UPV_nombre", "UPV_contratos", "UPV_importe", "UV_nombre",
        "UV_nif", "UV_contratos", "UV_importe", "UPV_NORM", "UV_NORM"
    ]).sort_values("SIMILITUD", ascending=False)
    cand.to_excel(FICHERO_REVISION, index=False)
    print(f"  -> {len(cand):,} candidatos escritos en {FICHERO_REVISION} "
          f"(marca S en MISMO_SN y vuelve a ejecutar)")

# ---------- merge final: RESUMEN proveedor-año ----------
for df in (upv, uv):
    df["GRUPO"] = df["NORM"].map(uf.find)

todo = pd.concat([upv[["GRUPO", "NORM"]], uv[["GRUPO", "NORM"]]])
nombre_grupo = todo.groupby("GRUPO")["NORM"].agg(moda)  # nombre representativo


def agregar(df, prefijo, extras):
    agg = {
        "nombre": ("PROVEEDOR", moda), "n_contratos": ("IMPORTE", "size"),
        "importe_total": ("IMPORTE", "sum"), "importe_medio": ("IMPORTE", "mean"),
        "nif": ("NIF", moda),
    }
    for c in extras:                       # Objeto, Tipo contrato, CONCEPTO ECONÓMICO, CENTRO DIRECTIVO
        agg[c] = (c, unicos)
    g = df.groupby(["GRUPO", "AÑO"]).agg(**agg)
    g.columns = [f"{prefijo}_{c}" for c in g.columns]
    return g


a = agregar(upv, "UPV", EXTRAS_UPV).drop(columns=["UPV_nif"])
b = agregar(uv, "UV", EXTRAS_UV)
m = a.join(b, how="outer").reset_index()
m["PROVEEDOR_CLAVE"] = m["GRUPO"].map(nombre_grupo)

for c in ["UPV_n_contratos", "UV_n_contratos", "UPV_importe_total", "UV_importe_total"]:
    m[c] = m[c].fillna(0)
m["UPV_n_contratos"] = m["UPV_n_contratos"].astype(int)
m["UV_n_contratos"] = m["UV_n_contratos"].astype(int)
m["EN_AMBAS"] = (m["UPV_n_contratos"] > 0) & (m["UV_n_contratos"] > 0)
m["TOTAL_N_CONTRATOS"] = m["UPV_n_contratos"] + m["UV_n_contratos"]
m["TOTAL_IMPORTE"] = m["UPV_importe_total"] + m["UV_importe_total"]

cols = ["PROVEEDOR_CLAVE", "AÑO", "EN_AMBAS",
        "UPV_nombre", "UPV_n_contratos", "UPV_importe_total", "UPV_importe_medio",
        "UPV_CENTRO DIRECTIVO", "UPV_CONCEPTO ECONÓMICO",
        "UV_nombre", "UV_nif", "UV_n_contratos", "UV_importe_total", "UV_importe_medio",
        "UV_Tipo contrato", "UV_Objeto",
        "TOTAL_N_CONTRATOS", "TOTAL_IMPORTE"]
m = m[cols].sort_values(["EN_AMBAS", "TOTAL_IMPORTE"], ascending=[False, False])

print(f"\nFilas (proveedor-año): {len(m):,}")
print(f"  solo UPV: {((m.UPV_n_contratos>0)&(m.UV_n_contratos==0)).sum():,}")
print(f"  solo UV : {((m.UPV_n_contratos==0)&(m.UV_n_contratos>0)).sum():,}")
print(f"  en ambas: {m.EN_AMBAS.sum():,}  ({m.loc[m.EN_AMBAS,'PROVEEDOR_CLAVE'].nunique():,} proveedores distintos)")
control = m["TOTAL_N_CONTRATOS"].sum() + sin_nombre
total = len(raw_upv) + len(raw_uv)
print(f"Control de contratos: {control:,} de {total:,} -> {'OK' if control == total else 'REVISAR (filas sin AÑO válido)'}")


# ---------- DETALLE: un contrato por fila, todas las columnas originales ----------
def detalle_contratos():
    u = raw_upv.copy()
    u["IMPORTE"] = a_numero(u["IMPORTE"])
    u = u.rename(columns={"EXPEDIENTE": "Identificador / EXPEDIENTE",
                          "TERCERO": "Nombre adjudicatario / TERCERO"})
    u.insert(0, "UNIVERSIDAD", "UPV")

    v = raw_uv.copy()
    for c in IMPORTES_UV:
        v[c] = a_numero(v[c])
    v["Fecha adjudicación"] = a_fecha(v["Fecha adjudicación"])
    v = v.rename(columns={"Identificador": "Identificador / EXPEDIENTE",
                          "Nombre adjudicatario": "Nombre adjudicatario / TERCERO"})
    v.insert(0, "UNIVERSIDAD", "UV")

    d = pd.concat([u, v], ignore_index=True)
    d["AÑO"] = pd.to_numeric(d["AÑO"], errors="coerce").astype("Int64")
    norm = d["Nombre adjudicatario / TERCERO"].map(normalizar_nombre)
    d.insert(1, "PROVEEDOR_CLAVE",
             norm.map(lambda n: nombre_grupo.get(uf.find(n)) if n in uf.p else None))
    # Garantía de que no se pierde ninguna columna de origen
    assert set(d.columns) == set(ORDEN_DETALLE), set(d.columns) ^ set(ORDEN_DETALLE)
    return d[ORDEN_DETALLE]


det = detalle_contratos()
print(f"Detalle: {len(det):,} contratos (UPV {len(raw_upv):,} + UV {len(raw_uv):,})")


# ---------- guardado ----------
def para_excel(df):
    """Excel interpretaría como fórmula un texto que empiece por '='."""
    df = df.copy()
    for c in df.columns:
        if pd.api.types.is_string_dtype(df[c]):
            df[c] = df[c].str.replace(r"^=", "'=", regex=True)
    return df


def guardar_xlsx(ruta, hojas):
    try:
        import xlsxwriter  # noqa: F401
        motor, kw = "xlsxwriter", {"options": {"strings_to_formulas": False,
                                               "strings_to_urls": False}}
    except ImportError:
        motor, kw = "openpyxl", {}
    with pd.ExcelWriter(ruta, engine=motor, engine_kwargs=kw,
                        datetime_format="dd/mm/yyyy", date_format="dd/mm/yyyy") as xw:
        for nombre, df in hojas.items():
            para_excel(df).to_excel(xw, sheet_name=nombre, index=False)
            ws = xw.sheets[nombre]
            if motor == "xlsxwriter":
                ws.freeze_panes(1, 0)
                ws.autofilter(0, 0, len(df), len(df.columns) - 1)
            else:
                ws.freeze_panes = "A2"
                ws.auto_filter.ref = ws.dimensions


m.to_csv(SALIDA_CSV, sep=";", decimal=",", index=False, encoding="utf-8-sig", float_format="%.2f")
print("Guardado:", SALIDA_CSV)
if SALIDA_DETALLE_CSV:
    det.to_csv(SALIDA_DETALLE_CSV, sep=";", decimal=",", index=False, encoding="utf-8-sig",
               float_format="%.2f", date_format="%d/%m/%Y")
    print("Guardado:", SALIDA_DETALLE_CSV)
if SALIDA_XLSX:
    hojas = {"proveedor_año": m}
    if SALIDA_DETALLE_CSV:
        hojas["detalle_contratos"] = det
    guardar_xlsx(SALIDA_XLSX, hojas)
    print("Guardado:", SALIDA_XLSX)
```

```python
import re
import unicodedata
import pandas as pd


def quitar_tildes(texto):
    texto = unicodedata.normalize("NFKD", texto)
    return "".join(
        caracter
        for caracter in texto
        if not unicodedata.combining(caracter)
    )


def normalizar_proveedor(nombre):
    if pd.isna(nombre):
        return nombre

    nombre = quitar_tildes(str(nombre).strip().upper())
    nombre = re.sub(r"[.,]", " ", nombre)
    nombre = " ".join(nombre.split())

    # Quita formas societarias terminales, incluyendo variantes sin puntuación.
    sufijos = (
        " S L N E", " S L L", " S C R L", " S R L",
        " S C P", " S COOP", " S L U", " S A U",
        " S C L", " S L", " S A", " S C",
        " L L C", " L T D", " B V",
    )

    anterior = None
    while nombre != anterior:
        anterior = nombre
        for sufijo in sufijos:
            if nombre.endswith(sufijo):
                nombre = nombre[:-len(sufijo)].strip()
                break

    return nombre


# Variantes claras que aparecen en tu listado.
# Las claves y los valores se comparan después de quitar tildes y puntuación.
EQUIVALENCIAS = {
    "EDEN SPRINGS ESPANA U": "EDEN SPRINGS ESPANA",
    "MEDIA MARKT VALENCIA COLON U": "MEDIA MARKT VALENCIA COLON",
    "EXCLUSIVAS DE PASCUAL Y FURIO": "EXCLUSIVAS PASCUAL Y FURIO",
    "ERASMUS ANTIQUARIAAT EN BOEKHANDEL BV":
        "ERASMUS ANTIQUARIAAT BOEKHANDEL B V",
    "ERASMUS BOEKHANDEL BV":
        "ERASMUS ANTIQUARIAAT BOEKHANDEL B V",
    "SARSTEDT U": "SARSTEDT",
}

EQUIVALENCIAS_NORMALIZADAS = {
    normalizar_proveedor(variante): normalizar_proveedor(canonico)
    for variante, canonico in EQUIVALENCIAS.items()
}


def normalizar_y_unificar(nombre):
    nombre_normalizado = normalizar_proveedor(nombre)
    return EQUIVALENCIAS_NORMALIZADAS.get(
        nombre_normalizado,
        nombre_normalizado,
    )


# Conserva el valor original y normaliza cada fila de la tabla de hechos.
df["PROVEEDOR_ORIGINAL"] = df["PROVEEDOR_CLAVE"]
df["PROVEEDOR_CLAVE"] = df["PROVEEDOR_CLAVE"].apply(normalizar_y_unificar)

# Revisa las variantes afectadas; no elimina registros.
cambios = (
    df.loc[
        df["PROVEEDOR_ORIGINAL"] != df["PROVEEDOR_CLAVE"],
        ["PROVEEDOR_ORIGINAL", "PROVEEDOR_CLAVE"],
    ]
    .drop_duplicates()
    .sort_values("PROVEEDOR_CLAVE")
)

cambios
```

```python
def sin_acentos(t):
    t = unicodedata.normalize("NFKD", t)
    return "".join(c for c in t if not unicodedata.combining(c))


RE_MOJIBAKE = re.compile("[ÃÂ][\u0080-\u00bf\u0152\u0153\u0160\u0161\u0178\u017d\u017e\u0192\u02c6\u02dc"
                         "\u2013\u2014\u2018-\u201e\u2020-\u2022\u2026\u2030\u2039\u203a\u20ac\u2122]")


def _reparar_par(m):
    for enc in ("latin-1", "cp1252"):
        try:
            return m.group(0).encode(enc).decode("utf-8")
        except (UnicodeEncodeError, UnicodeDecodeError):
            continue
    return m.group(0)


def arreglar_codificacion(texto):
    """'ESPAÃ&#145;A' -> 'ESPAÑA', 'VERSIÃ“N' -> 'VERSIÓN'."""
    t = html.unescape(str(texto))
    for _ in range(3):                      # hay texto doblemente codificado
        if "Ã" not in t and "Â" not in t:
            break
        nuevo = RE_MOJIBAKE.sub(_reparar_par, t)
        if nuevo == t:
            break
        t = nuevo
    return t


def limpiar_texto(t):
    if t is None or pd.isna(t) or str(t).strip() == "":
        return None
    t = re.sub(r"\s+", " ", re.sub(r"_x000[Dd]_", " ", arreglar_codificacion(t))).strip()
    return t or None


def mayus(t):
    return sin_acentos(t).upper() if isinstance(t, str) and t else ""


def limpiar_columna(serie):
    mapa = {v: limpiar_texto(v) for v in serie.dropna().unique()}
    return serie.map(mapa)


for col in ["Objeto_UV", "CENTRO DIRECTIVO", "CONCEPTO ECONÓMICO", "Tipo_Contrato_UV"]:
    antes = df[col].copy()
    df[col] = limpiar_columna(df[col])
    print(f"{col:<22} filas modificadas: {int((antes.fillna('') != df[col].fillna('')).sum()):>7,}")
print("Objeto_UV con mojibake restante:", int(df["Objeto_UV"].fillna("").str.contains("Ã|Â[^ ]|&#|_x000").sum()))
```

```python
ABREVIATURAS = {
    "ASOC": "ASOCIACION", "FUND": "FUNDACION", "FEDER": "FEDERACION", "UNIV": "UNIVERSIDAD",
    "UNIVERSITAT": "UNIVERSIDAD", "INVEST": "INVESTIGACION", "INVESTIG": "INVESTIGACION",
    "PROFES": "PROFESIONALES", "GRAL": "GENERAL", "ALACANT": "ALICANTE", "D": "DE",
    "FUNDACIO": "FUNDACION", "UNIVERSIDADE": "UNIVERSIDAD",
}
ARTICULOS = {"DE", "DEL", "LA", "LAS", "LOS", "EL", "Y", "I", "THE", "OF", "AND", "EN"}

FORMAS_LARGAS = [
    "SOCIEDAD ANONIMA UNIPERSONAL", "SOCIEDAD LIMITADA UNIPERSONAL", "SOCIEDAD LIMITADA PROFESIONAL",
    "SOCIEDAD COMANDITARIA SIMPLE", "SOCIEDAD COOPERATIVA VALENCIANA", "SOCIEDAD COOPERATIVA",
    "SOCIEDAD LIMITADA", "SOCIEDAD ANONIMA", "SOCIEDAD CIVIL", "SDAD COOP C LTDA", "SDAD COOP V",
    "SDAD COOP", "PUBLIC LIMITED COMPANY", "UNLIMITED COMPANY", "LIMITED COMPANY", "COMPANY",
    "S ?A ?U", "S ?L ?U", "S ?L ?P", "S ?L ?L", "S ?C ?P", "S ?C ?L", "S ?R ?L", "S ?P ?A", "S ?A ?S",
    "S ?A ?R ?L", "S ?COOP(?: ?V)?", "S ?C ?V", "U ?T ?E", "S ?A", "S ?L", "S ?C", "C ?B", "C ?L",
    "B ?V", "N ?V", "A ?G", "S ?E", "L ?T ?D ?A?", "LIMITED", "LIMITADA", "INC", "LLC", "GMBH", "SARL",
    "APS", "LDA", "KG", "CORP(?:ORATION)?", "R ?L", "SOCIETE A RESPONSABILITE LIMITEE", "R L UK",
]
FORMAS_UNA_LETRA = ["U", "P", "L", "E", "V"]
RE_FORMAS = re.compile(r"\s+(?:%s)$" % "|".join(FORMAS_LARGAS))
RE_FORMAS_1 = re.compile(r"\s+(?:%s)$" % "|".join(FORMAS_UNA_LETRA))
RE_SUCURSAL = re.compile(r"^(.+?)\s+(?:SUCURSAL|SUCCURSALE|NIEDERLASSUNG|BRANCH)\b.*$")


def normalizar_nombre(nombre):
    if nombre is None or pd.isna(nombre):
        return None
    t = mayus(arreglar_codificacion(nombre))
    t = re.sub(r"EU\d{9}", " ", t)                       # código de IVA pegado (IEEE...)
    t = re.sub(r"[^A-Z0-9 ]", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    base = t
    t = " ".join(ABREVIATURAS.get(w, w) for w in t.split())
    m = RE_SUCURSAL.match(t)
    if m:
        t = m.group(1)
    while True:                                          # formas jurídicas solo al final
        n = RE_FORMAS.sub("", t)
        if n == t and len(t.split()) >= 3:
            n = RE_FORMAS_1.sub("", t)
        if n == t or not n.strip():
            break
        t = n.strip()
    return t or base or None


def clave_palabras(norm):
    """'JAIME DOLZ CASANA' = 'DOLZ CASANA JAIME' (palabras ordenadas y sin artículos)."""
    pal = [w for w in norm.split() if w not in ARTICULOS]
    return " ".join(sorted(pal)) or norm


df["PROVEEDOR_ORIGINAL"] = df["Proveedor"]
mapa = {v: normalizar_nombre(v) for v in df["PROVEEDOR_ORIGINAL"].dropna().unique()}
df["PROVEEDOR_NORM"] = df["PROVEEDOR_ORIGINAL"].map(mapa)
print(f"{df['PROVEEDOR_ORIGINAL'].nunique():,} nombres -> {df['PROVEEDOR_NORM'].nunique():,} normalizados")
for t in ["AMAZON EU R L SUCURSAL EN ESPANA", "ELECNOR SERVICIOS Y PROYECTOS U",
          "IEEE THE INSTITUTE OF ELECTRICALEU372008451"]:
    print(f"  {t}  ->  {mapa.get(t)}")
```

```python
class UnionFind:
    def __init__(self, items):
        self.p = {i: i for i in items}

    def find(self, x):
        while self.p[x] != x:
            self.p[x] = self.p[self.p[x]]
            x = self.p[x]
        return x

    def union(self, a, b):
        if a in self.p and b in self.p:
            ra, rb = self.find(a), self.find(b)
            if ra != rb:
                self.p[rb] = ra
                return True
        return False

    def construir_claves(pares_extra=()):
        nombres = sorted(df["PROVEEDOR_NORM"].dropna().unique())
        uf = UnionFind(nombres)
        for fn in (lambda n: n.replace(" ", ""), clave_palabras):
            cubo = {}
            for n in nombres:
                k = fn(n)
                if k in cubo:
                    uf.union(cubo[k], n)
                else:
                    cubo[k] = n
        for a, b in pares_extra:
            uf.union(a, b)
        grupo = df["PROVEEDOR_NORM"].map(lambda n: uf.find(n) if n in uf.p else None)
        c = df.assign(GRUPO=grupo).groupby(["GRUPO", "PROVEEDOR_NORM"]).size().rename("n").reset_index()
        c = (c.sort_values(["GRUPO", "n", "PROVEEDOR_NORM"], ascending=[True, False, True])
            .drop_duplicates("GRUPO").set_index("GRUPO")["PROVEEDOR_NORM"])
        return grupo, grupo.map(c)


df["GRUPO"], df["PROVEEDOR_CLAVE"] = construir_claves()
print(f"{df['PROVEEDOR_NORM'].nunique():,} normalizados -> {df['PROVEEDOR_CLAVE'].nunique():,} claves")
```

```python
from rapidfuzz import fuzz, process

UMBRAL_FUZZY = 92
reps = df.groupby("GRUPO").agg(CLAVE=("PROVEEDOR_CLAVE", "first"), FILAS=("Importe", "size")).reset_index()
reps["BLOQUE"] = reps["CLAVE"].str[:1]
filas = []
for _, bloque in reps.groupby("BLOQUE"):
    b = bloque.reset_index(drop=True)
    if len(b) < 2:
        continue
    M = process.cdist(b["CLAVE"].tolist(), b["CLAVE"].tolist(), scorer=fuzz.token_sort_ratio,
                      score_cutoff=UMBRAL_FUZZY, dtype=np.uint8, workers=-1)
    for i, j in zip(*np.nonzero(np.triu(M, k=1))):
        filas.append({"SIMILITUD": int(M[i, j]), "NOMBRE_A": b.at[i, "CLAVE"], "FILAS_A": int(b.at[i, "FILAS"]),
                      "NOMBRE_B": b.at[j, "CLAVE"], "FILAS_B": int(b.at[j, "FILAS"]),
                      "NORM_A": b.at[i, "GRUPO"], "NORM_B": b.at[j, "GRUPO"]})
cand = pd.DataFrame(filas).sort_values(["SIMILITUD", "FILAS_A"], ascending=False).reset_index(drop=True)
print(len(cand), "candidatos")
cand[["SIMILITUD", "NOMBRE_A", "FILAS_A", "NOMBRE_B", "FILAS_B"]].head(40)
```
